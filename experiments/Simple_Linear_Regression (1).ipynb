```python
from google.colab import files
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
NVDA = files.upload()
NVDA = pd.read_csv("nvda_us_d.csv")
NVDA = NVDA.drop(columns=['Open','High','Low','Volume'])
NVDA
```

```python
train = NVDA['Close'].iloc[0:30]
X_train = train.iloc[0:-1].values.reshape(-1,1)
Y_train = train.iloc[1:].values
test = NVDA['Close'].iloc[30:]
X_test = test.iloc[0:-1].values.reshape(-1,1)
Y_test = test.iloc[1:].values
dates_test = NVDA['Date'].iloc[31:]
model = LinearRegression()
model.fit(X_train,Y_train)
prediction = model.predict(X_test)
model_accuracy = r2_score(Y_test,prediction)
model_accuracy = model_accuracy*100
model_graph = px.scatter(x=dates_test,y=Y_test)
model_graph.add_scatter(x=dates_test,y=prediction,mode='lines',name='Prediction Line')
model_graph.show()
```

```python
NVDA_new1 = {'Date':'2025-11-24','Close':182.55}
NVDA = pd.concat([NVDA,pd.DataFrame([NVDA_new1])],ignore_index=True)
next_day = NVDA['Close'].iloc[-1:].values.reshape(-1,1)
next_predict = model.predict(next_day)
print("The model's prediction for the next trading day(November 25 2025) is $",next_predict)

print("The accuracy of this model is",round(model_accuracy,2),"%")
```
